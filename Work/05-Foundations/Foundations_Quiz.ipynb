## Q5. Files and Paths (05)
Use `pathlib` and `with open(..., encoding='utf-8')`:
1. Write `Hello World` to `OUT / 'hello.txt'`.
2. **Append** a second line, `Goodbye World`, without erasing the first.
3. Read the whole file back and print it.
4. Print the file's `.name`, `.suffix`, and whether it `.exists()`.

```python
# Your code here
file_path = OUT / 'hello.txt'

# 1. Write the first line
with open(file_path, 'w', encoding='utf-8') as file:
    file.write('Hello World\n')

# 2. Append the second line
with open(file_path, 'a', encoding='utf-8') as file:
    file.write('Goodbye World\n')

# 3. Read the whole file
with open(file_path, 'r', encoding='utf-8') as file:
    content = file.read()

print(content)

# 4. Print file information
print("Name:", file_path.name)
print("Suffix:", file_path.suffix)
print("Exists:", file_path.exists())
```

## Q6. JSON (05)
Create a dictionary describing a course (name, number of credits, and a **list** of topics). Save it to `OUT / 'course.json'` with `indent=2`, read it back into a new variable, and print whether the two dictionaries are equal (`==`).

```python
# Your code here
course = {
    "name": "Computer Science",
    "credits": 3,
    "topics": ["Python", "Data Science", "Algorithms"]
}

# Save dictionary to JSON
json_path = OUT / 'course.json'

with open(json_path, 'w', encoding='utf-8') as file:
    json.dump(course, file, indent=2)

# Read JSON into a new variable
with open(json_path, 'r', encoding='utf-8') as file:
    loaded_course = json.load(file)

# Compare dictionaries
print(course == loaded_course)
```

## Q7. Data I/O with Pandas (06)
1. Create a small DataFrame (3–5 rows, your own data, at least one number column) and save it to `OUT / 'my_data.csv'` **without** the index.
2. Read it back with Pandas, and show the first rows with `.head()` and the summary statistics with `.describe()`.
3. Convert it to JSON at `OUT / 'my_data.json'` using `orient='records'`, and print the file's text.

## Q10. Timing (08)
Compare squaring one million numbers with a Python loop (or list comprehension) over a **list** and with NumPy on an **array**:
1. Create both inputs **before** timing, so the setup isn't measured.
2. Confirm that both approaches produce the same values.
3. Use `%timeit -o` for each, and print how many times faster NumPy is.

```python
# Your code here
# 1. Create inputs before timing
numbers_list = list(range(1_000_000))
numbers_array = np.arange(1_000_000)

# 2. Square the numbers using Python
def square_list(numbers):
    return [x ** 2 for x in numbers]

# Square the numbers using NumPy
def square_numpy(numbers):
    return numbers ** 2

# Confirm both approaches produce the same values
python_result = square_list(numbers_list)
numpy_result = square_numpy(numbers_array)

print("Results are equal:",
      np.array_equal(python_result, numpy_result))

# 3. Measure Python execution time
python_time = %timeit -o square_list(numbers_list)

# Measure NumPy execution time
numpy_time = %timeit -o square_numpy(numbers_array)

# Calculate how many times faster NumPy is
speedup = python_time.average / numpy_time.average

print(f"NumPy is approximately {speedup:.2f} times faster.")
```
